### 02 - Recursos (Importações)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sbn

from mlxtend.frequent_patterns import apriori, association_rules

sbn.set_theme(style="whitegrid")
```

### 03 - Implementação

```python
dataframe = pd.read_csv("./Data/itens_2026-11-05-grande.csv", sep=";", on_bad_lines="skip")

dataframe = dataframe.dropna(subset=["arquivo_xml", "xProd"])

basket = dataframe.groupby(["arquivo_xml", "xProd"])["xProd"].count().unstack().reset_index().fillna(0).set_index("arquivo_xml")

basket_encoded = (basket > 0).astype(bool)

# Notas Fiscais e Produtos Únicos.

display(basket_encoded.head())
```

### 04 - Apriori

```python
frequent_itemsets = apriori(basket_encoded, min_support=0.02, use_colnames=True)

frequent_itemsets = frequent_itemsets.sort_values(by="support", ascending=False)

# Combinações mais Frequentes.

display(frequent_itemsets.head(10))
```

### 05 - Association Rules

```python
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

rules = rules.sort_values(by="confidence", ascending=False)

# Regras de Associação.

display(rules[["antecedents", "consequents", "support", "confidence", "lift"]].head(10))
```

### 05 - Gráfico

```python
plt.figure(figsize=(10,6))

scatter = plt.scatter(rules["support"], rules["confidence"], c=rules["lift"], cmap="plasma", alpha=0.8, s=100)

plt.title("Regras de Associação (Confiança/Suporte)", fontsize=14)

plt.xlabel("Suporte (Frequência)", fontsize=12)
plt.ylabel("Confiança (Probabilidade Condicional)", fontsize=12)

color_bar = plt.colorbar(scatter)

color_bar.set_label("Lift (Força da Associação)", rotation=270, labelpad=15)

plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()

plt.show()
```
