### 02 - Recursos (Importações)

```python
import pandas as pd

print(f"Pandas Version: {pd.__version__}")
```

### 03 - Implementação

```python
file_path = "./Data/itens_2026-10-26.csv"

try:
  products_dataframe = pd.read_csv(file_path, delimiter=";", decimal=".")

  print("Dados carregados com sucesso.\n")

  display(products_dataframe)
except:
  print("Erro ao carregar os dados!")

  exit()
```

### 04 - Estrutura

```python
print("Primeiras Linhas:\n")

display(products_dataframe.head())

print(
  "\nFormato do DataFrame:\n\n",
  products_dataframe.shape,
  "\n"
)

print(
  "\nEstrutura do DataFrame:\n\n",
  products_dataframe.info()
)

print(
  "\nExtração Individual de Coluna:\n\n",
  products_dataframe["xProd"]
)

print(
  "\nExtração Múltipla de Colunas:\n\n",
  products_dataframe[["xProd", "vProd", "qCom"]]
)
```

### 05 - Filtragem

```python
# Filtragem Individual.

unitary_filter_values = products_dataframe[products_dataframe["vProd"] > 1000]

display(unitary_filter_values)

# Filtragem Múltipla.

multiple_filter = (products_dataframe["xProd"] == "GASOLINA COMUM") & (products_dataframe["qCom"] < 500)

multiple_filter_values = products_dataframe[multiple_filter]

display(multiple_filter_values)
```
